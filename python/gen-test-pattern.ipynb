```python
# helper functions
from PIL import Image
def get_hex(a:int) -> str:
    return hex(a)[2:].upper()

# return R, G, B, w, h 
def get_image_hex(file:str):
    img = Image.open(file)
    pixels = img.load()
    w, h = img.size
    # print(w,h)
    rhex = []
    ghex = []
    bhex = []
    for x in range(w):
        for y in range(h):
            R,G,B,gamma = pixels[x,y]
            rhex.append(get_hex(R))
            ghex.append(get_hex(G))
            bhex.append(get_hex(B))
    return(rhex,ghex,bhex,w,h)
```

```python
from PIL import Image

img = Image.open("noita-420x720.png")
pixels = img.load()
w, h = img.size
print(w,h)
rhex = []
ghex = []
bhex = []
for x in range(w):
    for y in range(h):
        R,G,B,gamma = pixels[x,y]
        rhex.append(get_hex(R))
        ghex.append(get_hex(G))
        bhex.append(get_hex(B))

with open("rhex.hex","w",newline='\n') as fd:
    for i in rhex:
        print(i,file=fd)
with open("ghex.hex","w",newline='\n') as fd:
    for i in ghex:
        print(i,file=fd)
with open("bhex.hex","w",newline='\n') as fd:
    for i in bhex:
        print(i,file=fd)
```
